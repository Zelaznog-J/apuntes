## Conceptos

Un dato vectorial combina una **geometría** (punto, línea o polígono) con **atributos** en una tabla. GeoPandas extiende pandas con una columna `geometry` y un sistema de referencia (CRS).

## Ejemplo

Una geometría puede escribirse como texto WKT (*Well-Known Text*):

````python
ciudades = {"Concepción": (-73.05, -36.83), "Santiago": (-70.65, -33.45)}

for nombre, (lon, lat) in ciudades.items():
    print(f"{nombre}: POINT ({lon} {lat})")
````

Con GeoPandas la misma información queda como un `GeoDataFrame`:

```python
import geopandas as gpd

gdf = gpd.GeoDataFrame(
    {"nombre": list(ciudades)},
    geometry=gpd.points_from_xy(*zip(*ciudades.values())),
    crs="EPSG:4326",
)
```
